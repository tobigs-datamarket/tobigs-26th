**Q2 [Target Leakage].** 데이터에 '판매 직전 납부한 취등록세액' 변수가 있고 `SalePrice`와 상관이 0.99라고 하자. 취등록세는 보통 판매가에 비례해 매겨진다. 이 변수를 피처에서 **제외해야 하는 이유**를 Target Leakage 관점에서 설명하라. (힌트: 집을 팔기 *전에* 알 수 있는 값인가?)
→ **답:** 취들록세액은 집이 판매된 후에 판매가를 기준으로 계산되는 값이므로 집을 팔기 전에는 알 수 없다. 따라서 `SalePrice`의 정보를 미리 포함한 Target Leakage가 발생하므로 피처에서 제외해야 한다.

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# 간단히 수치형 변수 몇 개만 사용
feats = ['GrLivArea','OverallQual','TotalBsmtSF','GarageCars','LotFrontage']
X = df[feats].copy()
y = np.log1p(df['SalePrice'])

# 1) 가장 먼저 Train/Test 분할  (random_state는 본인 학번 뒤 4자리로!)
# TODO: 아래 ______ 에 본인 학번 앞 4자리 (예: 2021학번이면 2021)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2026)

# 2) 결측 대체: Train 에서 fit, Test 에는 transform 만
imp = SimpleImputer(strategy='median')
X_train_imp = imp.fit_transform(X_train)          # fit_transform: Train 통계를 '학습'하며 변환
# TODO: Test 는 새로 fit 하지 말고, Train 이 학습한 값으로 transform 만 하라
X_test_imp  = imp.transform(X_test)

# 3) 스케일링도 동일 원칙 (Train 에서만 fit)
sc = StandardScaler()
X_train_sc = sc.fit_transform(X_train_imp)
# TODO: Test 스케일링 (transform 만)
X_test_sc  = sc.transform(X_test_imp)

print("완료 — Train:", X_train_sc.shape, "/ Test:", X_test_sc.shape)
print("Train 중앙값으로 채운 LotFrontage 대체값:", round(imp.statistics_[feats.index('LotFrontage')], 1))
```

→ **답:** Test 데이터는 미래에 들어올 미지의 데이터이므로 test데이터의 평균이나 중앙값을 사용하면 미래 데이터의 정보를 미리 사용하게 된 것이므로 데이터 누수가 발생하고 모델 성능에 영향을 준다.

**Q2.** 결측 대체·스케일링 같은 전처리보다 **Train/Test 분할을 반드시 먼저** 해야 하는 본인의 행동 원칙을, 데이터 누수 방지 측면에서 말해주세요.
→ **답:** 먼저 train/test를 분할하고 결측 대체·스케일링은 train 데이터로만 fit하고 test 데이터에는 transform만 적용한다. 이렇게 해야 test 데이터의 정보가 모델 학습과정에서 유입되는 데이터 누수를 방지할 수 있다.

이 과제를 풀며 GPT/Claude 등을 **어디에, 어떻게** 썼는지 기록한다. (안 썼으면 "사용 안 함")

답변

활용한 부분: 이해안가는 코드 질문
대화 내역:

그럼 그냥 transform을 한다는것은 train 데이터 값을 이용한다는 거야?

맞아. 정확히 이해했어.

transform()은 Train 데이터에서 이미 학습(fit)한 기준을 그대로 사용하는 것이야.

예를 들어 StandardScaler를 쓰면,

scaler = StandardScaler()

# Train에서 평균과 표준편차를 학습
scaler.fit(X_train)

여기서 내부적으로

평균(mean) = 50
표준편차(std) = 10

을 기억해 둔다.

그 다음

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

을 하면,

Train도 평균 50, 표준편차 10으로 변환
Test도 평균 50, 표준편차 10으로 변환

한다.

즉, Test 자신의 평균이나 표준편차는 절대 계산하지 않는다.

반면 아래처럼 하면 안 된다.

scaler.fit(X_test)      # ❌
# 또는
scaler.fit_transform(X_test)   # ❌

이렇게 하면 Test의 평균과 표준편차를 미리 알아버리므로 미래 데이터를 엿본 것(Data Leakage) 이 된다.

정리
fit → Train 데이터의 통계값(평균, 표준편차, 중앙값 등)을 학습
transform → 학습한 Train의 통계값을 그대로 적용
fit_transform → fit + transform을 한 번에 수행 (Train에서만 사용)

그래서 보통은

X_train = scaler.fit_transform(X_train)  # ✅
X_test  = scaler.transform(X_test)       # ✅

이 패턴을 항상 사용한다.
