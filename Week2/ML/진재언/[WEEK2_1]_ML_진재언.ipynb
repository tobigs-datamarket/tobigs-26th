### 1-1. ML 문제 정의 (수업 §2)

- 학습 전에 데이터를 train/test로 나누는 이유를 **본인의 언어로** 한 문장: train/test 분할이 없이 데이터를 학습하면 성능이 부풀려질 수 있으며 일반화를 할 수 없다.


### 1-2. 전처리와 데이터 분할

아래 TODO를 채우고, **각 주석의 빈칸에 '왜 이 단계가 필요한지'를 직접 적으세요.


````python
#LLM 사용
#age 분포 확인
plt.figure(figsize=(8, 4))
sns.histplot(titanic["age"].dropna(), bins=30, kde=True)
plt.axvline(titanic["age"].median(), color="red", linestyle="--", label=f"median: {titanic['age'].median():.1f}")
plt.axvline(titanic["age"].mean(), color="orange", linestyle="--", label=f"mean: {titanic['age'].mean():.1f}")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.legend()
plt.show()
````

````python
# 사용할 feature와 target 선택
features = ["pclass", "sex", "age", "fare"]
target = "survived"

df = titanic[features + [target]].copy()

# [빈칸] age의 결측치를 중앙값으로 채우는 이유 (평균이 아니라 중앙값을 쓰는 이유 포함): 위의 분포에서 알 수 있듯 age는 80세까지 있기에 평균을 사용하면 이러한 큰 값들에 영향을 받을 수 있어 중앙값을 사용하는 것이 더 적절하다.
df["age"] = df["age"].fillna(df["age"].median())

# [빈칸] sex를 0/1 숫자로 바꾸는 이유 (모델 입장에서 설명): 모델이 문자열 그 자체를 이해하지 못하므로 0과 1로 변환을 해주어 계산이 가능하도록 해야 한다. 
df["sex"] = (df["sex"] == "female").astype(int)

df = df.dropna()

X = df[features]
y = df[target]

# TODO: train/test를 8:2로 분할하세요. random_state=MY_SEED 사용
# [빈칸] stratify=y 옵션을 주는 이유: survived의 클래스 비율을 train과 test에 유지해야 할 필요가 있다.
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=MY_SEED,
    stratify=y
)

print(f"train: {X_train.shape}, test: {X_test.shape}")
print(f"train 생존율: {y_train.mean():.3f}, test 생존율: {y_test.mean():.3f}")
````

### 1-3. Gradient Descent 손계산 (수업 §3)

수업에서 "학습 = loss를 낮추는 방향으로 파라미터를 조금씩 갱신"이라고 했습니다. 이걸 **가장 작은 예제로 직접 계산**합니다.

**설정**: 데이터 3개 $(x, y) = (1,2), (2,4), (3,6)$, 모델 $\hat{y} = wx$ (절편 없음), loss는 MSE

$$L(w) = \frac{1}{3}\sum_{i=1}^{3}(y_i - wx_i)^2, \qquad \frac{\partial L}{\partial w} = -\frac{2}{3}\sum_{i=1}^{3} x_i(y_i - wx_i)$$

**문제**: $w_0 = 0$, learning rate $\eta = 0.1$에서 시작하여 **gradient descent를 3회 반복해 직접 계산**하고, **각 단계에서 $w$가 왜 그 방향으로, 왜 그만큼 움직였는지** 설명하세요. (계산기는 써도 되지만 코드로 먼저 답을 구하면 안 됩니다 — 아래 검증 셀은 손계산이 끝난 뒤에 실행)

| 반복 | 현재 $w$ | 각 데이터의 오차 $(y_i - wx_i)$ | gradient $\frac{\partial L}{\partial w}$ | 갱신된 $w$ |
|---|---|---|---|---|
| 1 | 0 | 2, 4, 6 | -18.67 | 1.87 |
| 2 | 1.87 | 0.13, 0.26, 0.39 | -1.21 | 1.99 |
| 3 | 1.99 | 0.01, 0.02, 0.03 | -0.09 | 2 |

**서술**: 매 반복마다 gradient의 크기가 어떻게 변했고, 그것이 $w$의 이동량과 어떤 관계인가? 이 과정이 언제, 왜 멈추게 될까? → gradient의 크기는 감소하는 것을 알 수 있었으며 $w$의 이동량은 반복 횟수가 증가함에 따라 이동량의 폭은 계속 감소하였다는 것을 알 수 있다. 이는 $w$가 true value에 가까워질수록 이동량이 작아져야 true value를 찾게 될 수 있을 것이므로 이동량의 폭이 감소한다는 것을 유추해볼 수 있다. 이 과정은 gradient가 0에 근사했을 때 멈출 것이다.


````python
# ===== 손계산 검증용 셀 (표를 다 채운 뒤에 실행하세요) =====
# 주의: 위에서 만든 target y(Series)를 덮어쓰지 않도록 여기서는 gx, gy 라는 별도 변수를 쓴다.
gx = np.array([1, 2, 3])
gy = np.array([2, 4, 6])

w = 0.0
lr = 0.1

for step in range(1, 4):
    # TODO: 위 수식대로 gradient를 코드로 옮기세요
    grad = (-2/len(gx)) * np.sum(gx*(gy-w*gx))   # 힌트: -2/len(gx) * np.sum(...)

    # [빈칸] gradient의 '반대 방향'으로 이동하는 이유: MSE, 즉 loss를 줄이기 위해
    w = w - lr * grad

    loss = np.mean((gy - w * gx) ** 2)
    print(f"step {step}: grad = {grad:+.4f}, w = {w:.4f}, loss = {loss:.4f}")

# 손계산 결과와 출력이 일치하는지 확인하고, 다르면 어디서 틀렸는지 찾아 적으세요: 거의 차이가 없을 확인할 수 있다. 
````

### 1-4. 두 모델 학습과 비교

수업에서 "모든 알고리즘은 (f의 형태 / loss / 찾는 방법)에 대한 서로 다른 답"이라고 했습니다. 형태가 전혀 다른 두 모델을 **같은 데이터, 같은 조건**에서 학습시켜 비교합니다.

- **모델 A**: Logistic Regression (선형 + sigmoid)
- **모델 B**: Decision Tree (재귀 분할)


````python
# 과정 확인 장치: train과 test 성능을 모두 기록합니다
results = {}

# TODO: 모델 A — LogisticRegression을 학습시키세요 (max_iter=1000)
model_a = LogisticRegression(max_iter=1000)
model_a.fit(X_train, y_train)

# TODO: 모델 B — DecisionTreeClassifier를 학습시키세요 (random_state=MY_SEED)
model_b = DecisionTreeClassifier(random_state=MY_SEED)
model_b.fit(X_train, y_train)

for name, model in [("Logistic Regression", model_a), ("Decision Tree", model_b)]:
    # [빈칸] train 정확도와 test 정확도를 '둘 다' 기록하는 이유 (수업 §3의 개념과 연결해서): train의 정확도와 test의 정확도의 차이가 크다면 overfitting을 의심해볼 수 있다.
    train_acc = accuracy_score(y_train, model.predict(X_train))
    test_acc = accuracy_score(y_test, model.predict(X_test))
    results[name] = (train_acc, test_acc)
    print(f"{name:20s} | train acc: {train_acc:.4f} | test acc: {test_acc:.4f}")
````

### 1-5. 평가지표 직접 유도

수업에서 "Confusion Matrix 하나에서 4개 지표가 전부 유도된다"고 했습니다. **sklearn의 지표 함수를 쓰기 전에**, TP/TN/FP/FN으로부터 직접 계산해 봅니다. (모델 A 기준)


````python
y_pred = model_a.predict(X_test)

cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", cm)

# sklearn의 confusion_matrix 배치: [[TN, FP], [FN, TP]]
TN, FP = cm[0]
FN, TP = cm[1]
print(f"TP={TP}, TN={TN}, FP={FP}, FN={FN}")

# TODO: 아래 4개 지표를 TP/TN/FP/FN '만으로' 직접 계산하세요 (sklearn 함수 사용 금지)
my_accuracy  = (TP+TN) / (TP+TN+FP+FN)
my_precision = TP / (TP+FP)   # [빈칸] precision의 분모에 들어가는 것과 그 의미: 분모는 true positive와 false positive의 합으로, 모델이 양성으로 예측한 것 중 실제로 양성인 비율을 나타낸다.
my_recall    = TP / (TP+FN)   # [빈칸] recall의 분모에 들어가는 것과 그 의미: 분모는 true positive와 false negative의 합으로, 실제로 양성인 것 중 모델이 양성으로 예측한 비율을 나타낸다.
my_f1        = 2*(my_precision*my_recall) / (my_precision+my_recall)   # 힌트: precision과 recall의 조화평균

# 검증: sklearn과 비교 (통과하지 못하면 수식을 다시 확인)
assert np.isclose(my_accuracy,  accuracy_score(y_test, y_pred))
assert np.isclose(my_precision, precision_score(y_test, y_pred))
assert np.isclose(my_recall,    recall_score(y_test, y_pred))
assert np.isclose(my_f1,        f1_score(y_test, y_pred))
print("✅ 4개 지표 모두 sklearn과 일치")
````

### 1-6. 결과 해석 (필수 서술)

아래 4개 질문에 **본인의 실험 결과 수치를 근거로** 답하세요. (수치 없이 일반론만 쓰면 감점)

1. **왜 이러한 결과가 나왔는가?** — 두 모델의 test 정확도는 각각 얼마였고, 이 데이터에서 그 정도 성능이 나온 이유를 feature 관점에서 추측하면? → 로지스틱의 test 정확도는 0.7877. 트리의 test 정확도는 0.7654였다. 사용했던 feature가 "pclass", "sex", "age", "fare" 4개뿐이였다는 점에서 이 정도 성능이 나왔다는 것을 알 수 있고, 적절한 변수 추가 및 제거는 더 높은 정확도를 줄 수도 있을 것이다.
2. **두 모델의 성능 차이는 어디에서 발생했는가?** — train acc와 test acc의 '차이'가 두 모델에서 어떻게 달랐는가? 이를 모델 구조(선형 경계 vs 재귀 분할)의 관점에서 해석하면? → 로지스틱은 단순한 선형 경계를 만들기에 train과 test 정확도 사이에 큰 차이가 없었으나 트리는 보다 더 정확한 분류를 위해 재귀 분할을 한다는 점에서 overfitting을 의심해볼 수 있다. 실제로 train과 test 사이에 0.2121이라는 큰 차이가 있음을 볼 수 있다. 
3. **실험 조건을 변경하면 결과가 어떻게 달라지는가?** — `MY_SEED`를 다른 값으로 바꿔 1-2 ~ 1-4를 다시 실행해 보고, 어떤 수치가 얼마나 흔들렸는지 기록. 이 흔들림을 줄이는 방법으로 수업에서 배운 것은? → 시드를 2027에서 123으로 바꾸었더니 트리 모델의 정확도가 더 큰 차이를 보였다(train acc: 0.9846 | test acc: 0.7263). 수업에서는 validation set을 따로 추가하는 방식을 통해 이러한 문제를 해결할 수 있음을 배웠다. 
4. **예상과 다른 결과가 나왔다면 그 원인은 무엇인가?** — 과제를 시작하기 전 예상과 달랐던 지점 하나와, 그 원인에 대한 본인의 가설: → 그래도 트리 모델이 로지스틱보다는 예측을 더 잘 할 줄 알았으나 오히려 로지스틱이 test data에 대해 우수한 성능을 보였다는 점에서 트리 모델에서 overfitting이 발생했음을 유추해볼 수 있다. 


````python
MY_SEED_2 = 123  # TODO: 랜덤한 숫자
np.random.seed(MY_SEED_2)

#  Titanic 데이터를 사용
titanic_2 = sns.load_dataset("titanic")

# 사용할 feature와 target 선택
features = ["pclass", "sex", "age", "fare"]
target = "survived"

df = titanic[features + [target]].copy()

# [빈칸] age의 결측치를 중앙값으로 채우는 이유 (평균이 아니라 중앙값을 쓰는 이유 포함): 위의 분포에서 알 수 있듯 age는 80세까지 있기에 평균을 사용하면 이러한 큰 값들에 영향을 받을 수 있어 중앙값을 사용하는 것이 더 적절하다.
df["age"] = df["age"].fillna(df["age"].median())

# [빈칸] sex를 0/1 숫자로 바꾸는 이유 (모델 입장에서 설명): 모델이 문자열 그 자체를 이해하지 못하므로 0과 1로 변환을 해주어 계산이 가능하도록 해야 한다. 
df["sex"] = (df["sex"] == "female").astype(int)

df = df.dropna()

X = df[features]
y = df[target]

# TODO: train/test를 8:2로 분할하세요. random_state=MY_SEED 사용
# [빈칸] stratify=y 옵션을 주는 이유: survived의 클래스 비율을 train과 test에 유지해야 할 필요가 있다.
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=MY_SEED_2,
    stratify=y
)

print(f"train: {X_train.shape}, test: {X_test.shape}")
print(f"train 생존율: {y_train.mean():.3f}, test 생존율: {y_test.mean():.3f}")
````

````python
# 과정 확인 장치: train과 test 성능을 모두 기록합니다
results = {}

# TODO: 모델 A — LogisticRegression을 학습시키세요 (max_iter=1000)
model_a = LogisticRegression(max_iter=1000)
model_a.fit(X_train, y_train)

# TODO: 모델 B — DecisionTreeClassifier를 학습시키세요 (random_state=MY_SEED)
model_b = DecisionTreeClassifier(random_state=MY_SEED_2)
model_b.fit(X_train, y_train)

for name, model in [("Logistic Regression", model_a), ("Decision Tree", model_b)]:
    # [빈칸] train 정확도와 test 정확도를 '둘 다' 기록하는 이유 (수업 §3의 개념과 연결해서): train의 정확도와 test의 정확도의 차이가 크다면 overfitting을 의심해볼 수 있다.
    train_acc = accuracy_score(y_train, model.predict(X_train))
    test_acc = accuracy_score(y_test, model.predict(X_test))
    results[name] = (train_acc, test_acc)
    print(f"{name:20s} | train acc: {train_acc:.4f} | test acc: {test_acc:.4f}")
````

## 2. 심화 과제 (선택)

### 2-1. Overfitting 직접 재현하기

 overfitting/underfitting curve를 **본인 데이터로 직접 그립니다.** Decision Tree의 `max_depth`를 1부터 15까지 바꿔가며 train/test 정확도를 기록하세요.


````python
depths = range(1, 16)
train_scores, test_scores = [], []

for d in depths:
    # TODO: max_depth=d인 DecisionTreeClassifier를 학습시키고
    #       train/test 정확도를 각 리스트에 추가하세요
    model_c = DecisionTreeClassifier(max_depth=d)
    model_c.fit(X_train, y_train)

    train_scores.append(model_c.score(X_train, y_train)) #DecisionTreeClassifier.score()로 정확도 출력
    test_scores.append(model_c.score(X_test, y_test))

plt.figure(figsize=(8, 5))
plt.plot(depths, train_scores, marker="o", label="train")
plt.plot(depths, test_scores, marker="s", label="test")
# Colab 기본 환경엔 한글 폰트가 없어 제목/라벨은 영문으로 둔다(한글이면 □로 깨짐)
plt.xlabel("max_depth (model complexity)")
plt.ylabel("accuracy")
plt.legend(); plt.grid(alpha=0.3)
plt.show()

# 해석 (필수):
# - 그래프에서 underfitting 구간과 overfitting 구간은 각각 어디인가? → underfitting 구간은 max_depth가 1~3 정도인 구간이고, overfitting 구간은 max_depth가 7 이상인 구간이다.
# - 본인 데이터 기준 최적의 max_depth와 그렇게 판단한 근거는? → train과 test간의 차이가 가장 적었던 max_depth가 4이므로 최적의 max_depth는 4로 판단하였다. 
# - 이 그래프가 수업의 "train loss만 낮추는 건 쉽다"는 문장과 어떻게 연결되는가? → max_depth가 커질수록 train 정확도는 계속 올라가지만 test 정확도는 특정 구간부터 감소하기 시작한다. max_depth를 만약에 17 이상으로 하였다면 train의 정확도는 거의 1에 가깝게 만들 수 있음을 의미하지만, 결국 이는 test loss까지 낮추는 건 어렵다는 것을 의미한다.
````
